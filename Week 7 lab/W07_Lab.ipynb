## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
raw_line=" Srv-01 , used=87 , total=100 "

line=raw_line.strip()#remove whitespaces

hostname_part,rest=line.split(",",1) #separates hostname from remaining string

hostname=hostname_part.strip().lower()#lowercase and cleaned hostname

record={"hostname":hostname}#starting a dictionary with hostname

for piece in rest.split(","): #loop through the remaining string
    key,value=piece.split("=") #separates key and value from piece
    record[key.strip()]=float(value.strip()) #clean and store in dictionary

print(record)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """Converts raw data into a structured dictionary"""
    line=raw_line.strip()#remove whitespaces

    hostname_part,rest=line.split(",",1) #separates hostname from remaining string

    hostname=hostname_part.strip().lower()#lowercase and cleaned hostname

    record={"hostname":hostname}#starting a dictionary with hostname

    for piece in rest.split(","): #loop through the remaining string
        key,value=piece.split("=") #separates key and value from piece
        record[key.strip()]=float(value.strip()) #clean and store in dictionary

    return record #returns the completed dictionary
print(parse_line("SRV-02,used=120,total=100"))
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
with open("messy_lines.txt") as f: #open the text file in read mode
    # text=f.read()
    for line in f:    #read each line of text file
        ln=parse_line(line) #convert the line into a dictionary
        host=ln["hostname"]
        percent=(ln["used"]/ln["total"])*100 #calculate percentage to deduce "status"
        
        if percent>=100:
            status="OVER LIMIT"
        elif percent>=90:
            status="WARNING" 
        else:
            status="OK"

        print(host,status)
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**
```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}
Could not parse: srv-08,used=unknown,total=100
Could not parse: srv-09,note=a=b,used=1,total=2
```

````python
# C2
lines=[
    "srv-07,used=50,total=100",
    "srv-08,used=unknown,total=100",
    "srv-09,note=a=b,used=1,total=2"
]
for line in lines:
    try:#catch errorcode 
        record=parse_line(line)
        print(record)
    except ValueError:
        print("Could not parse:",line) #displays which records are error inducing
````
